# Midterm Probability Review

Review questions for the test over probability. First, let's initialize our **perm()** function which evaluates permutations for us.

```r
perm <- function(n, k) {
    return(choose(n,k) * factorial(k))}
```

## Review Problems

1. Let $A$ and $B$ be two events in a sample space $S$ such that $P(A) = 0.5$, $P(B) = 0.4$, and $P(A \cap B) = 0.2$. What is the probability of the union $A \cup B$?

2. If $A$ and $B$ are non-empty events with $P(A) > 0$ and $P(B) > 0$, and they are **mutually exclusive**, can they also be **independent**?

3. A card is drawn from a standard $52$-card deck. Given that the card drawn is a face card, what is the probability that it is a King?

4. A committee of $3$ people is to be chosen from a group of $10$ people. How many different committees can be formed?

5. If $A$ and $B$ are independent events with $P(A) = 0.4$ and $P(B) = 0.5$, what is the conditional probability $P(A \mid B)$?

6. How many distinct $3$-letter arrangements (permutations) can be formed using the letters A, B, C, D, and E if no letter is repeated in a single arrangement?

7. Let $A$ and $B$ be two events in a sample space such that $P(A) = 0.4$, $P(B) = 0.5$, and $P(A \cup B) = 0.7$. What is $P(A \cap B)$?

8. A committee of $3$ people is to be formed from a group of $5$ men and $4$ women. How many different committees can be formed that consist of exactly $2$ men and $1$ woman?

9. If $P(A) = 0.6$, $P(B) = 0.5$, and $P(A \mid B) = 0.8$, what is the conditional probability $P(B \mid A)$?

10. A fair, four-sided die (tetrahedron) has its faces numbered $1$, $2$, $3$, and $4$. The die is rolled once, and the sample space is $S = \{1, 2, 3, 4\}$, where each outcome is equally likely. Let the events be defined as:
- $A = \{1, 2\}$
- $B = \{1, 3\}$
- $C = \{1, 4\}$

a) Show that the events $A, B$ and $C$ are pairwise independent.

b) Determine whether the events $A, B,$ and $C$ are mutually independent. 

11. A standard state vehicle license plate consists of 2 letters followed by 4 digits. If repetition of both letters and digits is allowed, how many unique license plates can be created? (Assume 26 letters and 10 digits available).

12. A custom registration code requires 3 distinct letters followed by 2 distinct digits. If no letter or digit can be repeated anywhere within that same code, how many different codes are possible?

13. A bookshelf has space for 7 distinctly different novels. In how many different linear arrangements can these 7 novels be placed side-by-side?

14. A club with 10 members needs to elect a President, a Vice President, and a Treasurer. Assuming no member can hold more than one office, how many ways can these three positions be filled?

15. A science department needs to select an unassigned committee of 4 professors from a pool of 12 eligible professors. How many different committees can be formed?

16. A project team of 4 people is to be chosen from a group of 6 men and 5 women. How many ways can the team be formed if it must consist of exactly 2 men and 2 women?

17. There are 9 horses competing in a race. If you randomly place an "exacta" bet (meaning you guess which horse finishes in exactly 1st place and which horse finishes in exactly 2nd place), what is the probability that your bet is correct?

18. A student club has 10 members, including Alice and Bob. The club needs to randomly select a President and a Vice President from its members. What is the probability that Alice is chosen as President and Bob is chosen as Vice President?

19. If you randomly shuffle the letters of the word "MATH", what is the probability that the letters end up in the exact correct order spelling "MATH"?

20. A high-tech manufacturer packs components into boxes of 20. A quality control protocol randomly selects 5 components from a box for testing. If more than 1 component is found to be defective, the entire box is rejected.Suppose a specific box contains exactly 4 defective components. What is the probability that this box passes the inspection?

21. A corporate board consists of 7 men and 6 women. A special task force of 5 members must be chosen randomly.What is the probability that the committee will contain a strict majority of women, given that two specific board members (one man and one woman) refuse to serve on the committee together?

22. A secure government server cluster contains 30 nodes divided into three security tiers:15 Standard Nodes10 Secure Nodes5 Critical NodesAn advanced persistent threat (APT) actor manages to bypass the firewall and randomly compromises exactly 8 nodes before being locked out. What is the probability that the hacker compromised exactly 4 Standard Nodes and at least 2 Critical Nodes?

23. A network router transmits data packets across a noisy channel. Due to interference, each packet has an independent probability of $p = 0.15$ of being corrupted during transmission. Suppose a message consists of a batch of $n = 12$ packets. Let $X$ be the number of corrupted packets in the batch.

- Calculate the probability that exactly $2$ packets are corrupted.
- Calculate the probability that at most $2$ packets are corrupted.

24. A manufacturing plant produces microchips, and the probability of a microchip being defective is $p = 0.05$. Quality control engineers test the chips one by one in independent trials until they find exactly $3$ defective chips. Let $X$ represent the total number of microchips tested.

- Find the exact probability that the third defective chip is found on the twentieth test.
- Calculate the expected number of tests required to find $3$ defective chips.

25. An internet router drops data packets independently with a constant probability of \(p = 0.04\). Let \(X\) be the number of packets sent up to and including the first dropped packet.Identify the distribution of \(X\) and state its support.Find the probability that the first dropped packet is exactly the 10th packet sent.Calculate the probability that the router successfully transmits at least 20 packets before a packet is dropped (i.e., \(P(X > 20)\)).

26. The length of a customer service phone call (in minutes) follows an exponential distribution with a mean service time of $\mu = 8$ minutes. 

- What is the probability that a randomly selected call lasts longer than $12$ minutes?
- Given that a call has already lasted $5$ minutes, what is the conditional probability that the total call length will exceed $17$ minutes?

27. A charity raffle sells tickets for $\$5$ each. Out of $500$ tickets sold, one ticket wins a grand prize of $\$500$, three tickets win second-place prizes of $\$100$ each, and ten tickets win third-place prizes of $\$20$ each. The remaining tickets win nothing. Let $X$ be the net profit of a single ticket buyer. Find the expected value and standard deviation of $X$.

28. An electronics retailer offers a one-year laptop protection plan for $\$40$. Historical data shows a $5\%$ chance a client files a minor damage repair claim costing the company $\$200$, and a $1\%$ chance they file a total-loss replacement claim costing the company $\$800$. Otherwise, no claims are made. Let $Y$ be the insurance company's net profit per policy. Find $E(Y)$ and $SD(Y)$.

29. A small box contains $5$ billiard balls: $3$ are red and $2$ are blue. Two balls are randomly drawn from the box without replacement. Let $R$ be the total number of red balls collected. Find the expected value and standard deviation of $R$.

30. An early-stage tech investor decides to fund a startup. Financial projections outline three clear outcomes over a 3-year timeline: a $20\%$ chance the startup fails completely resulting in a loss of $\$1,000$, a $50\%$ chance it breaks even returning $\$0$ net gain, and a $30\%$ chance it succeeds wildly resulting in a net profit of $\$3,000$. Let $V$ be the investment return. Find $E(V)$ and $\sigma_V$.
