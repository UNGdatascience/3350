# P06d: DiscreteProbabilityFunctions

We have four scenarios which are quite common in probability problem-solving. 
1. Binomial
2. Geometric
3. Hypergeometric
4. Negative Binomial

The examples below will be drawn from all four of them. Run the cell below to initialize our permutations function **perm**.

```r
perm <- function(n, k) {
    return(choose(n,k) * factorial(k))}
```

## Example 1: Interviews

A job applicant receives a callback for an interview with probability $p = 0.14$ per application submitted. What is the probability that the applicant has to submit more than 5 applications before receiving their first callback?

## Example 2: Rainy Days

A local weather analyst estimates a $30\%$ chance of rain on any given day during a $14$-day spring period. Assuming independence across days, find the probability that it rains on at least $3$ but no more than $6$ days.

## Example 3: Broadway Cards

A standard deck of $52$ playing cards contains $k = 20$ Broadway Cards, named for the values needed for a **Broadway Straight**. A player is dealt a hand of $n = 5$ cards without replacement. (a) What is the probability of drawing exactly 3 Broadway Cards? (b) What is the probability of drawing at least 3 Broadway Cards?

## Example 4: Sales

A salesperson makes cold calls where each call has an independent $p = 0.18$ probability of closing a sale. What is the probability that their $3$rd successful sale occurs on exactly their $10$th cold call?

## Example 5: Quality Control

A quality control engineer inspects a random sample of $n = 12$ manufactured components. Historical data shows that $8\%$ of components are defective. Find the probability that exactly 2 components in the sample are defective.

## Example 6: Committee Selection

A university department consists of $N = 20$ faculty members, of which $K = 8$ are tenured. A dean randomly selects a committee of $n = 6$ faculty members without replacement. Find the probability that a majority (at least 4) of the selected committee members are tenured.

## Example 7: Marine Biology

A marine biologist tags deep-sea fish. The probability of capturing a specific tagged species per net deployment is $p = 0.20$. The researcher needs to collect $r = 4$ specimens. Calculate the probability the researcher will complete the data collection in $7$ net deployments, or fewer.

## Example 8: Multiple Choice

A student takes a 10-question multiple-choice quiz where each question has 4 options ($p = 0.25$). If the student guesses randomly on every question, what is the probability they pass by answering at least 7 questions correctly?

## Example 9: Sports Series

Two players are playing a match where the winner must be the first to win $r = 4$ games. If Player A wins each game with probability $p = 0.60$ independently, what is the probability that Player A wins the match on the 6th game?
