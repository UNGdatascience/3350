## <span style="color:red;font-weight:bold;font-size:1.1em">Data Checks and What we are Looking For</span>

William Seally Gossett created the $t$-statistic specifically for research scenarios with small sample sizes, and this is where the $t$ shines. When working in an area requiring sample sizes of

<center><span style="color:blue;font-size:1.25em">$n\leq 50$</span></center>

the $t$-statistic will provide better accuracy than the **brittle** $z$-stastistic. The $t$-statistic is **robust** and resists the penalties when small violation of the assumptions occur as, in the real world, is often the case.

The **assumptions** of the $t$-test and other parametric tests are as follows:
1. Normality
2. Independence
3. Homogeneity of the Variances

#### To avoid difficulties in tests where the samples have different standard deviations:
We want to avoid **sharply unequal** sample sizes. Hence, we try to ensure sample size ratio of $$\textbf{largest}:\textbf{smallest}$$ is not greater than $$2:1$$
