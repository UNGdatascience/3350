## Categorical Data Analysis with $\chi^2$

The goals of <span style = 'color:blue; font-weight:bold'>linear regression</span> and the <span style = 'color:blue; font-weight:bold'>$\chi^2$ test of independence</span> are very similar:

- A **linear regression** determines if a relationship exists between 2 <span style = 'color:red; font-weight:bold'>numeric</span> variables.
- A **$\chi^2$ test of independence** determines if a relationship exists between 2 <span style = 'color:red; font-weight:bold'>category</span> variables.

## Data Verification

To use $\chi^2$ Test of Independence, we must have no more than 20\% low cell counts in the **expected matrix** where "low cell count" is defined to be **less than five.**

The examples below begin with the **observed data matrix** being constructed directly, while the third example will need to calculate the values in the data matrix using **xtabs()**.

## Example 1: Learning Styles

A psychologist questions whether preference of learning method differs by gender. A group survey is conducted the results of which are shown below. Test whether a relationship exists between gender and learning method.

    <th></th>
    <th style="text-align: center">Male</th>
    <th style="text-align: center">Female</th> 
  </tr>
  <tr>
    <td style="text-align: center"><b>Visual</b></td>
    <td style="text-align: center">25</td>
    <td style="text-align: center">14</td>

```r
data <- matrix(c(25, 24, 41, 14, 42, 34), nrow = 3)
data
```

```r
result <- chisq.test(data)
result$expected
```

**Data Check**

No values in the expected matrix are less than 5, so these data are appropriate for $\chi^2$ procedures.

**Hypotheis**

- $H_0 : $ Proportions of learning styles are **independent** of biological sex.
- $H_a : $ Proportions of learning styles are **dependent** upon biological sex.

```r
result
```

**Reporting Out**

Because $p = 0.0.1313 < 0.05 = \alpha$, we reject the null.  Evidence suggests that the proportion of learning styles depends upon biological sex.

**Hypothesis**

- $H_0 : $ Perecentage of moviegoers buying snacks is **independent** of type of movie seen.
- $H_a : $ Perecentage of moviegoers buying snacks is **dependent** upon type of movie seen.

```r
res
```

**Reporting out**

Since $p = 1.358\times 10^{-5} < 0.05 =\alpha$, we reject the null. Thus, evidence suggests that whether or not a patron buys snacks depends upon the type of movie they are going to see.

## Example 3: Accept the Date

Using the **AccDate** varible in the **personality** data set, test at the $0.05$ level of significance whether Yes/No responses to the dating question depends upon biological sex.

```r
pers <- read.csv('https://faculty.ung.edu/rsinn/data/personality.csv')
```

```r
xtabs(~Sex + AccDate, data = pers)
```

```r
obs2 <- matrix(c(28,28,46,27), nrow = 2)
obs2
```

```r
rz <- chisq.test(obs2)
rz$expected
```

**Data Checks**

No observed matrix values less than 5. OK.

**Hypothesis**

- $H_0 : $ Vars independent
- $H_a : $ Vars dependent

```r
rz
```

Since $p= 0.193 > 0.05 = \alpha$, we fail to reject the null. We thus have no evidence that the Yes/No responses depend upon biological sex.
