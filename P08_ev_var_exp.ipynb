$$f(x) = \begin{cases} 3x^2 & 0 \leq x \leq 1 \\ 0 & \text{otherwise} \end{cases}$$

### Example 4: Insurance Risk & Portfolio Expected Value

A major claim costs the insurer $\$50,000$. A minor claim costs the insurer $\$5,000$.The annual policy premium charged to the customer is $\$2,000$.Let $X$ be the random variable representing the net payout by the insurer for a single policy (where a claim payout is a positive expense and the premium is revenue).Construct the probability distribution table for the insurer's net payout $X$.Calculate the expected net payout $E(X)$ per policy for the insurer.Determine the minimum premium $P_{\text{min}}$ the insurer must charge to break even on average.
